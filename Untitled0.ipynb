**Описание данных**
 * User Id - id пользователя

 * Region - регион визита

 * Device - устройство, с которого производится вход на сайт

 * Channel - рекламный канал перехода на сайт

 * Session Start - время начала пользовательской сессии (входа на сайт)

 * Session End - время окончания пользовательской сессии (уход с сайта)

 * SessionDurationSec - длительность сессии в секундах

 * Session Date - дата визита

 * Month - месяц визита

 * Day - день дня недели визита

 * Hour of day - час (от 0 до 24) визита

 * Order Dt - дата покупки

 * Revenue - сумма покупки

 * Payment Type - тип оплаты

 * Promo code - использование промокода на скидку 10% (есть-1, нет-0).

Добавим столбец "time_of_day" с указанием времени суток визита("утро", "день", "вечер", "ночь")

```python
#столбец с указанием времени суток визита
def time_of_day_of_the_visit(session):
    time = int(session[11:13])
    if time >= 6 and time < 10:
        return "утро"
    elif time >= 10 and time < 17:
        return "день"
    elif time >= 17 and time < 22:
        return "вечер"
    else:
        return "ночь"

df['time_of_day'] = df['session_start'].apply(time_of_day_of_the_visit) # время суток
df.head(3)
```

Регрессионное моделирование

```python
# Линейная модель, которая предсказывает продажи исходя из двух факторов: региона и рекламного канала, так как именно от них зависит выручка
df_without_nan = df.dropna()

X = df_without_nan[['region', "channel"]]
y = df_without_nan['total_amount']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42)

preprocessor = ColumnTransformer(
    transformers=[
        ('region_and_channel', OneHotEncoder(), ['region', "channel"])
    ],
    remainder='passthrough'
)

model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(np.round(y_pred))
```
